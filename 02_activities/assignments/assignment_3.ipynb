#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

**Question:**
- Do you notice any patterns or relationships between the different features? How might these patterns help in distinguishing between different species?

The scatter plot matrix has positive correlations between phenolic compounds like total_phenols vs flavanoids and negative relationships like color_intensity vs hue. Distinct 2-3 cluster groupings are visible across many feature pairs like proline, flavanoids, and total_phenols. Outliers appear across multiple plots, especially with proline showing extreme values that separate from main clusters. Thus, indicating that the three wine varieties have unique chemical signatures. The clear separation between groups with minimal overlap suggests that classification algorithms can easily draw decision boundaries.

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

We standardize our data before clustering to ensure that all features contribute equally to the distance calculations, preventing any single feature from dominating the results due to its scale. In clustering, distances between data points are critical. Features like proline ranges ~400-1600 and nonflavanoid_phenols ranges ~0.1-0.7 have vastly different scales. Without standardization, proline would dominate the analysis simply due to its larger numerical range, while smaller-scale features would have minimal impact on cluster formation. Standardizing ensures all features contribute equally to the grouping process. This is critical when feature ranges differ significantly, such as the wine dataset where chemical measurements span different orders of magnitude. Standardization transforms all features to the same scale mean=0, standard deviation=1, allowing the clustering algorithm to detect meaningful patterns across all chemical properties rather than being biased toward high-variance features.

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# First making a copy of the original data so that we don’t overwrite it while pre-processing and perform K-means clustering:
clustered_wine_data = wine_df.copy()
clustered_wine_data['Cluster'] = KMeans(n_clusters=3, random_state=123, n_init=10).fit(scaled_features).labels_

# Display first few rows with the 'Cluster' column:
print("First few rows with cluster labels:")
print(clustered_wine_data.head())
print(f"\nCluster distribution:")
print(clustered_wine_data['Cluster'].value_counts().sort_index())


# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

We use the Elbow Method to plot the Within-Cluster Sum of Squared Distances (WSSD) against different k values. As k increases, WSSD decreases, but the "elbow point" shows where additional clusters don't significantly improve quality. The process invloves looping through k values, running K-means for each, and use scikit-learn's .inertia_ attribute to get WSSD automatically. Its helpful because it provides visual identification of optimal k, balances clustering quality vs. complexity, prevents overfitting, and offers objective guidance over arbitrary selection. The elbow point represents the best trade-off between cluster cohesion and avoiding excessive fragmentation; typically showing the natural groupings in the data.

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Calculate the mean and setting the seed 
mean_color_intensity = wine_df['color_intensity'].mean()
np.random.seed(123)
# Generate 10000 bootstrap samples
bootstrap_samples = [wine_df.sample(frac=1, replace=True)['color_intensity'].mean() for _ in range(10000)]

# Calculate 90% confidence interval (5th and 95th percentiles)

lower_bound = np.percentile(bootstrap_samples, 5)
upper_bound = np.percentile(bootstrap_samples, 95)

# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

Bootstrapping approximates the sampling distribution of our sample mean without needing the full population, showing us the natural variability in sample means and helping assess how reliable our single sample estimate is.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

The confidence interval quantifies the uncertainty around our point estimate. A 90% confidence interval gives us a range where we can be 90% confident the true population mean lies. It transforms the abstract concept of "sampling variability" into concrete bounds, helping us understand the precision of our estimate and make informed decisions about the reliability of our conclusions.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

Yes, the mean appears stable and reliable. With 178 samples (large sample size), we get a tight confidence interval. If the CI is narrow relative to the mean value and the standard error is small, this indicates our sample provides a precise estimate of the population mean color intensity.
